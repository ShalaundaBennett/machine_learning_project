```python
# standard imports
import pandas as pd
import numpy as np

# Do not change the line of code below
pd.set_option('display.max_columns', 20)
```

```python
# Step 2: Load each file as a separate DataFrame.

bike_2011 = pd.read_csv("bike_per_hour_2011.csv")
bike_2012 = pd.read_csv("bike_per_hour_2012.csv")
weather = pd.read_csv("weather.csv")
```

```python
# Step 3: Fill missing temp and atemp values for February 2012 only.
# In this dataset, yr = 1 means 2012 and mnth = 2 means February.
feb_2012 = (weather["yr"] == 1) & (weather["mnth"] == 2)

# Calculate the February 2012 mean values with Pandas and round as instructed.
temp_mean = weather.loc[feb_2012, "temp"].mean().round(2)
atemp_mean = weather.loc[feb_2012, "atemp"].mean().round(4)

# Fill only the missing February 2012 temp and atemp values.
weather.loc[feb_2012, "temp"] = weather.loc[feb_2012, "temp"].fillna(temp_mean)
weather.loc[feb_2012, "atemp"] = weather.loc[feb_2012, "atemp"].fillna(atemp_mean)
```

```python
# Step 4: Combine the 2011 and 2012 bike rental DataFrames.
bike = pd.concat([bike_2011, bike_2012], ignore_index=True)
```

```python
# Step 5: Merge the bike rental data with the weather data.
# These five columns together identify the matching hourly record.
bike_weather = pd.merge(
    bike,
    weather,
    on=["yr", "season", "mnth", "day", "hr"],
    how="inner"
)
```

```python
# Step 6: Add total_count by adding casual and registered rentals.
bike_weather["total_count"] = bike_weather["casual"] + bike_weather["registered"]
```

```python
# Step 7: Delete dteday because the same date information is already in other columns.
bike_weather = bike_weather.drop(columns=["dteday"])
```

```python
# Step 8: Rename columns to make them easier to understand.
bike_weather = bike_weather.rename(columns={
    "yr": "year",
    "mnth": "month",
    "hr": "hour"
})
```

```python
# Step 9: Arrange the columns in the required order.
column_order = [
    "year", "season", "month", "day", "hour", "holiday", "weekday", "workingday",
    "weathersit", "temp", "atemp", "hum", "windspeed", "casual", "registered", "total_count"
]

bike_final = bike_weather[column_order]
```

```python
# Step 10 and 11: Sort the DataFrame and reset the index.
bike_final = bike_final.sort_values(by=["year", "month", "day", "hour"])
bike_final = bike_final.reset_index(drop=True)
```
